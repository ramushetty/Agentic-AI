```python
@dataclass
class BankContext:
    account_tier: str
```

```python
from langchain.tools import tool, ToolRuntime

@tool
def greet_customer(runtime: ToolRuntime[BankContext]) -> str:
    """Greet the customer based on their account tier."""

    tier = runtime.context.account_tier

    if tier == "premium":
        return (
            "Welcome to NovaBank Premium! "
            "Enjoy your exclusive banking benefits."
        )

    return "Welcome to NovaBank! You are a Standard customer."
```

```python


agent = create_agent(
    model="openai:gpt-4.1-mini",
    tools=[greet_customer],
    context_schema=BankContext,
    system_prompt=(
        "You are NovaBank's banking assistant. "
        "Always call greet_customer when greeting a customer."
    )
)
```

```python
# Standard customer
standard_response = agent.invoke(
    {
        "messages": [
            {"role": "user", "content": "Hello!"}
        ]
    },
    context=BankContext(account_tier="standard")
)
```

```python
print(standard_response)
```

```python
# Premium customer
premium_response = agent.invoke(
    {
        "messages": [
            {"role": "user", "content": "Hello!"}
        ]
    },
    context=BankContext(account_tier="premium")
)

print(premium_response["messages"][-1].content)
```

```python

from langchain.agents import create_agent
from langchain.agents.middleware import HumanInTheLoopMiddleware
from langchain.tools import tool
from langgraph.checkpoint.memory import InMemorySaver
from langgraph.types import Command


# 1. Define the transfer tool

@tool
def transfer_funds(to_account: str, amount: float) -> str:
    """Transfer money to another bank account."""

    print("\n*** TRANSFER TOOL EXECUTED ***")
    print(f"Destination account: {to_account}")
    print(f"Amount transferred: ₹{amount}")

    return (
        f"Successfully transferred ₹{amount} "
        f"to account {to_account}."
    )


# 2. Configure HITL middleware

hitl = HumanInTheLoopMiddleware(
    interrupt_on={
        "transfer_funds": True
    }
)


# 3. Create agent with checkpointer

agent = create_agent(
    model="openai:gpt-4.1-mini",
    tools=[transfer_funds],
    middleware=[hitl],
    checkpointer=InMemorySaver(),
    system_prompt=(
        "You are NovaBank's banking assistant. "
        "Use transfer_funds when the customer requests a transfer. "
        "Never claim a transfer succeeded unless the tool executes."
    )
)


# 4. Configure thread

config = {
    "configurable": {
        "thread_id": "transfer-123"
    }
}


# 5. First invocation: trigger interrupt

result1 = agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": (
                    "Transfer ₹10000 to account ACC123."
                )
            }
        ]
    },
    config=config,
    version="v2"
)


# 6. Inspect the interrupt

print("\n--- HUMAN REVIEW REQUIRED ---")

for interrupt in result1.interrupts:
    print(interrupt.value)


# 7. Resume with EDIT decision

result2 = agent.invoke(
    Command(
        resume={
            "decisions": [
                {
                    "type": "edit",
                    "edited_action": {
                        "name": "transfer_funds",
                        "args": {
                            "to_account": "ACC123",
                            "amount": 5000.0
                        }
                    }
                }
            ]
        }
    ),
    config=config,
    version="v2"
)


# 8. Print final result

print("\n--- FINAL RESULT ---")

for message in result2.messages:
    if hasattr(message, "content"):
        print(message.content)
```

```python
import re

from langchain.agents import create_agent
from langchain.agents.middleware import PIIMiddleware, PIIMatch


# 1. Custom detector
def detect_novabank_account(content: str) -> list[PIIMatch]:
    """Detect NovaBank account numbers: NB- followed by 8 digits."""

    matches: list[PIIMatch] = []

    pattern = r"\bNB-\d{8}\b"

    for match in re.finditer(pattern, content):
        matches.append({
            "type": "novabank_account",
            "value": match.group(),
            "start": match.start(),
            "end": match.end(),
        })

    return matches


# 2. Configure PII middleware
pii_middleware = PIIMiddleware(
    "novabank_account",
    detector=detect_novabank_account,
    strategy="mask",
    apply_to_input=True,
)


# 3. Create agent
agent = create_agent(
    model="openai:gpt-4.1-mini",
    tools=[],
    middleware=[pii_middleware],
    system_prompt="You are NovaBank's banking assistant."
)


# 4. Test with an account number
result = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "Please check my account NB-12345678."
        }
    ]
})


# 5. Verify the message was masked
sanitized_message = result["messages"][0].content

print("Sanitized message:", sanitized_message)

assert "NB-12345678" not in sanitized_message

print("SUCCESS: Account number was masked!")
```

```python

import random
import time

from langchain.agents import create_agent
from langchain.agents.middleware import ToolRetryMiddleware
from langchain.tools import tool


# 1. Create a flaky banking API tool

@tool
def check_live_balance(account_id: str) -> str:
    """Check the live balance of a bank account."""

    print(f"Checking balance for {account_id}...")

    # Simulate 50% API failure
    if random.random() < 0.5:
        print("Connection failed!")
        raise ConnectionError("Banking API connection failed")

    print("API request successful!")

    return f"Account {account_id} balance is ₹75,000."


# 2. Configure retry middleware

retry_middleware = ToolRetryMiddleware(
    max_retries=3,
    backoff_factor=1.0,
    initial_delay=1.0,
    max_delay=5.0,
    jitter=False,
    retry_on=(ConnectionError,),
)


# 3. Create agent

agent = create_agent(
    model="openai:gpt-4.1-mini",
    tools=[check_live_balance],
    middleware=[retry_middleware],
    system_prompt=(
        "You are NovaBank's banking assistant. "
        "Use check_live_balance when asked for an account balance."
    )
)


# 4. Run multiple times

for run in range(1, 11):

    print(f"\n========== RUN {run} ==========")

    try:
        result = agent.invoke({
            "messages": [
                {
                    "role": "user",
                    "content": "Check the balance of account NB12345678."
                }
            ]
        })

        print("Agent response:")
        print(result["messages"][-1].content)

    except Exception as error:
        print("Final failure:", error)
```

```python

from langchain.agents import create_agent
from langchain.agents.middleware import (
    SummarizationMiddleware,
    ToolCallLimitMiddleware,
    ModelCallLimitMiddleware,
)
from langchain.tools import tool


# 1. Define banking tools

@tool
def check_balance(account_id: str) -> str:
    """Check the balance of a bank account."""

    print(f"Checking balance for {account_id}")

    return f"Account {account_id} balance is ₹75,000."


@tool
def transfer_funds(to_account: str, amount: float) -> str:
    """Transfer money to another account."""

    print(f"Transferring ₹{amount} to {to_account}")

    return f"Successfully transferred ₹{amount} to {to_account}."


# 2. Configure summarization middleware

summarization = SummarizationMiddleware(
    model="openai:gpt-4.1-mini",
    trigger={"messages": 12},
    keep=("messages", 6),
)


# 3. Configure tool call limits

# Overall tool limit: maximum 5 calls per run
global_tool_limit = ToolCallLimitMiddleware(
    run_limit=5,
)

# Specific tool limit: maximum 1 transfer per run
transfer_limit = ToolCallLimitMiddleware(
    tool_name="transfer_funds",
    run_limit=1,
)


# 4. Configure model call limit

model_limit = ModelCallLimitMiddleware(
    run_limit=6,
    exit_behavior="end",
)


# 5. Create agent with all three middleware

agent = create_agent(
    model="openai:gpt-4.1-mini",
    tools=[
        check_balance,
        transfer_funds,
    ],
    middleware=[
        summarization,
        global_tool_limit,
        transfer_limit,
        model_limit,
    ],
    system_prompt=(
        "You are NovaBank's banking assistant. "
        "Use the available tools to help customers. "
        "Never claim an operation succeeded unless its tool confirms it."
    ),
)


# 6. Test normal operation

response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": (
                "Check the balance of account NB12345678 "
                "and explain the result."
            ),
        }
    ]
})

print("\nAgent response:")
print(response["messages"][-1].content)


# 7. Test a normal transfer

response2 = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": (
                "Transfer ₹5000 to account ACC123."
            ),
        }
    ]
})

print("\nTransfer response:")
print(response2["messages"][-1].content)
```
